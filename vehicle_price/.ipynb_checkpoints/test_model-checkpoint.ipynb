# Test Saved Vehicle Price Prediction Model

```python
import joblib
import pandas as pd
import numpy as np

# Load the saved model and scaler
model = joblib.load('vehicle_model.joblib')
scaler = joblib.load('scaler.joblib')
```

```python
# Create a sample input (example)
# Format: [year, km_driven, fuel_encoded, seller_type_encoded, transmission_encoded, owner_encoded]
sample_input = [[2020, 30000, 2, 0, 1, 0]]

# Scale the input
sample_input_scaled = scaler.transform(sample_input)

# Make prediction
predicted_price = model.predict(sample_input_scaled)

print(f"Predicted price: ₹{predicted_price[0]:,.2f}")
```

### Feature Reference Guide

For the sample input, use these encodings:

- `fuel_encoded`: Petrol=0, Diesel=1, CNG=2, LPG=3
- `seller_type_encoded`: Individual=0, Dealer=1, Trustmark Dealer=2
- `transmission_encoded`: Manual=0, Automatic=1
- `owner_encoded`: First=0, Second=1, Third=2, Fourth & Above=3

```python
# Try multiple predictions
test_cases = [
    [2022, 10000, 0, 1, 1, 0],  # New automatic petrol car from dealer
    [2018, 50000, 1, 0, 0, 1],  # Older manual diesel car from individual
    [2020, 25000, 2, 2, 0, 0]   # Mid-age manual CNG car from trustmark dealer
]

# Scale and predict
test_cases_scaled = scaler.transform(test_cases)
predictions = model.predict(test_cases_scaled)

# Display results
for i, (case, pred) in enumerate(zip(test_cases, predictions)):
    print(f"\nCase {i+1}:")
    print(f"Year: {case[0]}")
    print(f"KM Driven: {case[1]:,}")
    print(f"Predicted Price: ₹{pred:,.2f}")
```
