```python
import joblib
import pandas as pd

# Load the saved model and scaler
model = joblib.load('vehicle_model.joblib')
scaler = joblib.load('scaler.joblib')

# Define feature names
feature_names = ['year', 'km_driven', 'fuel_encoded', 'seller_type_encoded', 
                 'transmission_encoded', 'owner_encoded']

# Create sample data using pandas DataFrame
sample = pd.DataFrame([[2020, 30000, 2, 0, 1, 0]], columns=feature_names)

# Display input data
print("Input Data:")
print(sample)
print("\nFeature Details:")
print("Year:", sample['year'].values[0])
print("Kilometers Driven:", sample['km_driven'].values[0])
print("Fuel Type (encoded):", sample['fuel_encoded'].values[0])
print("Seller Type (encoded):", sample['seller_type_encoded'].values[0])
print("Transmission (encoded):", sample['transmission_encoded'].values[0])
print("Owner (encoded):", sample['owner_encoded'].values[0])

# Make prediction
sample_scaled = scaler.transform(sample)
predicted_price = model.predict(sample_scaled)

print("\nPrediction Result:")
print(f"Predicted price: ₹{predicted_price[0]:,.2f}")
```
